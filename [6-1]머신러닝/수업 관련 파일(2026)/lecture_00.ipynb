# 머신러닝 수업을 위한 가상환경 설정 가이드

## 프로젝트 디렉토리 설정

1. **프로젝트 디렉토리 생성**  
   머신러닝 수업을 위한 작업 디렉토리를 생성합니다.
   - 예: `C:\machine_learning`

## A. Python 3.13 설치하기

1. **Python 설치**  
   - Windows 사용자: Python 공식 웹사이트에서 **Python 3.13.0** 설치 파일을 다운로드합니다.
2. **환경 변수 설정**  
   - 설치 과정에서 **"Add python.exe to PATH"** 옵션을 선택하여 환경 변수에 추가합니다.

## B. 가상환경 생성 및 활성화

1. **가상환경 생성**  
   - 명령 프롬프트(cmd)를 열고, 프로젝트 디렉토리(`C:\machine_learning`)로 이동한 후 다음 명령을 실행합니다:
     ```
     python -m venv ml  # "ml"은 가상환경 이름입니다.
     ```

2. **가상환경 활성화**  
   - 프로젝트 디렉토리에서 다음 명령을 실행하여 생성한 가상환경을 활성화합니다:
     ```
     ml\Scripts\activate
     ```
     또는
     ```
     ml\Scripts\activate.bat
     ```
   - 활성화가 완료되면 `(ml)`로 시작하는 표시가 명령 프롬프트에 나타납니다.

## C. 가상환경에 필요한 패키지 설치

1. **pip 및 setuptools 업그레이드**  
   - 최신 패키지 설치를 위해 필요에 따라 아래 명령을 실행하여 `pip`와 `setuptools`를 업데이트합니다:
     ```
     python -m pip install --upgrade pip setuptools
     ```

2. **필수 패키지 설치**  
   - 가상환경 내에서 다음 명령을 실행하여 머신러닝에 필요한 패키지를 설치합니다:
     ```
     pip install jupyterlab numpy scipy matplotlib pandas
     ```

3. **가상환경 종료**  
   - 가상환경에서 기본 시스템 환경으로 돌아가려면 다음 명령을 실행합니다:
     ```
     deactivate
     ```
     만약 `deactivate` 명령이 작동하지 않는다면, 다음 명령을 실행해 보세요:
     ```
     ml\Scripts\deactivate.bat
     ```

## D. JupyterLab Variable Inspector 설치 (추가 기능)

1. **Variable Inspector 설치**  
   - JupyterLab에서 변수들을 쉽게 확인할 수 있도록 Variable Inspector 확장을 설치합니다. 가상환경 내에서 다음 명령을 실행합니다:
     ```
     pip install lckr-jupyterlab-variableinspector
     ```

2. **활성화 방법**
   - 가상환경에서 `jupyter lab` 명령을 실행하여 **JupyterLab**을 시작합니다.
   - **JupyterLab**이 실행되면, 상단 메뉴의 **"View"** 탭에서 "Open Variable Inspector"를 선택하여 변수 창을 엽니다.
   - 또는 노트북 셀 영역에서 마우스 오른쪽 버튼을 클릭하고 "Open Variable Inspector"를 선택할 수 있습니다.



## 주피터 노트북 커맨드 모드 단축키

- **ESC**: **커맨드 모드**로 진입 (셀 수정 모드에서 벗어남).  
- **ENTER**: **수정 모드(Edit Mode)**로 돌아가 현재 셀을 수정 가능.

## 커맨드 모드 단축키
- **A**: 현재 셀 **위에** 새로운 셀 추가.
- **B**: 현재 셀 **아래에** 새로운 셀 추가.
- **D + D**: `D`를 연속 두 번 눌러 **현재 셀 삭제**.
- **M**: 선택한 셀을 **Markdown 셀**로 변환.
- **Y**: 선택한 셀을 **Code 셀**로 변환.
- **CTRL + B**: 왼쪽 **파일 탐색기**를 숨기거나 표시하여 화면을 더 크게 사용.
- **SHIFT + M**: 두 개의 셀을 **하나의 셀로 병합**.
- **CTRL + SHIFT + -**: 커서 위치를 기준으로 **두 개의 셀로 분할**.
