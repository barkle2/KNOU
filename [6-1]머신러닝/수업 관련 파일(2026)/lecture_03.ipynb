```python
def stats(species):
    data = iris[iris['Species'] == species]
    data_num = data.iloc[:, 0:4].values
    return np.mean(data_num, axis=0), np.std(data_num, axis=0)
```

```python
iris_num = iris.iloc[:, 0:4]
iris_mean = np.mean(iris_num, axis=0)
iris_std = np.std(iris_num, axis=0)
iris_standard = (iris_num - iris_mean) / iris_std
print(iris_standard.head())
```

```python
print(iris_standard.agg(['mean', 'std', 'min', 'max']))
```

```python
iris_min = np.min(iris_num, axis=0)
iris_max = np.max(iris_num, axis=0)
iris_minmax = (iris_num - iris_min) / (iris_max - iris_min)
print(iris_minmax.head())
```

```python
print(iris_minmax.agg(['mean', 'std', 'min', 'max']))
```

```python
covariance_matrix = np.cov(iris_standard, rowvar=False)
eigen_values, eigen_vectors = np.linalg.eig(covariance_matrix)
print(eigen_values)
print(eigen_vectors)
```

```python
print(np.corrcoef(iris_standard, rowvar=False))
```

```python
pc_4 = eigen_vectors
transformed_4 = np.dot(iris_standard, pc_4)
print(transformed_4[:7])
```

```python
tf_4_pd = pd.DataFrame(transformed_4, columns=['PC1', 'PC2', 'PC3', 'PC4'])
print(tf_4_pd.agg(['mean', 'std', 'min', 'max']))
```

```python
print(np.corrcoef(tf_4_pd, rowvar=False))
```

```python
pc_2 = eigen_vectors[:, :2]
transformed_2 = np.dot(iris_standard, pc_2)
print(transformed_2[:7])
```

```python
print(np.corrcoef(transformed_2, rowvar=False))
```

```python
import matplotlib.pyplot as plt

setosa = iris[iris['Species'] == 'setosa']
versicolor = iris[iris['Species'] == 'versicolor']
virginica = iris[iris['Species'] == 'virginica']
```

```python
plt.scatter(setosa['Petal.Length'], setosa['Petal.Width'],
            color='red', label='Setosa')
plt.scatter(versicolor['Petal.Length'], versicolor['Petal.Width'],
            color='green', label='Versicolor')
plt.scatter(virginica['Petal.Length'], virginica['Petal.Width'],
            color='blue', label='Virginica')

plt.title('Iris Petal Length vs Width')
plt.xlabel('Petal Length')
plt.ylabel('Petal Width')

plt.legend()
plt.show()
```
