## 문제 1. 이진 분류를 위한 LDA 방법론 유도

### 1-1. Between-class scatter matrix $S_B$가 왜 $(\mathbf{m}_2 - \mathbf{m}_1)(\mathbf{m}_2 - \mathbf{m}_1)^T$로 되는지 설명하시오. (5점)

#### 출발점: 성능지표(PI)의 분자

LDA는 투영 방향 $\mathbf{w}$를 찾기 위해 다음 성능지표를 최대화한다.

$$\text{PI}_\text{LDA}(\mathbf{w}) = \frac{(m_2 - m_1)^2}{N_1 s_1^2 + N_2 s_2^2}$$

여기서 $m_i = \mathbf{w}^T \mathbf{m}_i$는 클래스 $i$의 평균 벡터 $\mathbf{m}_i$를 축 $\mathbf{w}$에 투영한 **스칼라 값**이다.

---

#### 분자를 행렬 이차형식으로 변환

**Step 1.** 투영된 평균 차이를 전개한다:

$$m_2 - m_1 = \mathbf{w}^T\mathbf{m}_2 - \mathbf{w}^T\mathbf{m}_1 = \mathbf{w}^T(\mathbf{m}_2 - \mathbf{m}_1)$$

**Step 2.** 이 값은 스칼라이므로, 제곱은 자기 자신과의 곱이다:

$$(m_2 - m_1)^2 = \left[\mathbf{w}^T(\mathbf{m}_2 - \mathbf{m}_1)\right]^2 = \left[\mathbf{w}^T(\mathbf{m}_2 - \mathbf{m}_1)\right]\left[\mathbf{w}^T(\mathbf{m}_2 - \mathbf{m}_1)\right]$$

**Step 3.** 스칼라에 대해 $a = a^T$이므로:

$$= \mathbf{w}^T(\mathbf{m}_2 - \mathbf{m}_1) \cdot \left[(\mathbf{m}_2 - \mathbf{m}_1)^T\mathbf{w}\right]$$

**Step 4.** 곱셈 순서를 정리하면 ($\mathbf{w}$: $D\times1$, $(\mathbf{m}_2-\mathbf{m}_1)$: $D\times1$):

$$(m_2 - m_1)^2 = \mathbf{w}^T \underbrace{(\mathbf{m}_2 - \mathbf{m}_1)(\mathbf{m}_2 - \mathbf{m}_1)^T}_{D \times D \text{ 행렬}} \mathbf{w} = \mathbf{w}^T S_B \mathbf{w}$$

---

#### 결론

$$\boxed{S_B \triangleq (\mathbf{m}_2 - \mathbf{m}_1)(\mathbf{m}_2 - \mathbf{m}_1)^T}$$

이는 두 클래스 평균 차이 벡터의 **외적(outer product)**으로, 원래의 $D$차원 공간에서 클래스 간 평균 분리의 방향과 크기를 인코딩하는 랭크-1의 양반정치(PSD) 행렬이다. 이 정의 덕분에 성능지표의 분자를 $\mathbf{w}^T S_B \mathbf{w}$라는 이차형식(quadratic form)으로 표현하여, 분모의 $S_W$와 통일된 형태로 라그랑주 최적화를 수행할 수 있게 된다.

### 1-2. Within-class scatter matrix $S_W$를 유도하시오. (5점)

#### 출발점: 성능지표(PI)의 분모

성능지표의 분모는 두 클래스의 **투영된 데이터의 분산 합** $(N_1 s_1^2 + N_2 s_2^2)$이다.

$$s_i^2 = \frac{1}{N_i}\sum_{n \in C_i}(y_n - m_i)^2, \quad y_n = \mathbf{w}^T\mathbf{x}_n, \quad m_i = \mathbf{w}^T\mathbf{m}_i$$

---

#### 분모를 행렬 이차형식으로 변환

**Step 1.** $N_i$를 곱하면 합산 형태가 된다:

$$N_1 s_1^2 + N_2 s_2^2 = \sum_{i=1}^{2}\sum_{n \in C_i}(y_n - m_i)^2$$

**Step 2.** $y_n = \mathbf{w}^T\mathbf{x}_n$, $m_i = \mathbf{w}^T\mathbf{m}_i$를 대입한다:

$$= \sum_{i=1}^{2}\sum_{n \in C_i}\left(\mathbf{w}^T(\mathbf{x}_n - \mathbf{m}_i)\right)^2$$

**Step 3.** $\mathbf{w}^T(\mathbf{x}_n - \mathbf{m}_i)$는 스칼라이므로, 1-1과 동일하게 이차형식으로 변환한다:

$$\left(\mathbf{w}^T(\mathbf{x}_n - \mathbf{m}_i)\right)^2 = \mathbf{w}^T(\mathbf{x}_n - \mathbf{m}_i)(\mathbf{x}_n - \mathbf{m}_i)^T\mathbf{w}$$

**Step 4.** $\mathbf{w}$는 $n$, $i$와 무관하므로 합산 기호 밖으로 꺼낸다:

$$N_1 s_1^2 + N_2 s_2^2 = \mathbf{w}^T \underbrace{\left[\sum_{i=1}^{2}\sum_{n \in C_i}(\mathbf{x}_n - \mathbf{m}_i)(\mathbf{x}_n - \mathbf{m}_i)^T\right]}_{S_W} \mathbf{w} = \mathbf{w}^T S_W \mathbf{w}$$

---

#### 결론

$$\boxed{S_W \triangleq \sum_{i=1}^{2}\sum_{n \in C_i}(\mathbf{x}_n - \mathbf{m}_i)(\mathbf{x}_n - \mathbf{m}_i)^T = S_1 + S_2}$$

### 1-3. LDA의 성능지표(목적함수)에 관하여 기술하고, 왜 효과적인지 직관적으로 설명하시오. (5점)

#### 성능지표 정의

$$\boxed{\text{PI}_\text{LDA}(\mathbf{w}) = \frac{(m_2 - m_1)^2}{N_1 s_1^2 + N_2 s_2^2} = \frac{\mathbf{w}^T S_B \mathbf{w}}{\mathbf{w}^T S_W \mathbf{w}}}$$

이 형태는 **Rayleigh quotient(레이리 몫)**라 불리며, $\mathbf{w}$에 임의의 상수 $c$를 곱해도 지표값이 변하지 않는 성질을 가진다. 따라서 $\mathbf{w}$의 방향만이 의미를 가진다.

---

#### 직관적 설명

| 상황 | 분자 $(m_2-m_1)^2$ | 분모 $N_1s_1^2+N_2s_2^2$ | PI | 분류 결과 |
|:---|:---:|:---:|:---:|:---:|
| 평균은 멀지만 분산이 큼 | 크다 | 크다 | 중간 | 겹침 발생 |
| 평균이 가깝고 분산이 작음 | 작다 | 작다 | 중간 | 분리 어려움 |
| **평균은 멀고 분산이 작음** | **크다** | **작다** | **크다** | **명확히 분리** |

PI가 클수록 두 클래스의 투영된 분포가 서로 멀리 떨어지면서 각자 뭉쳐 있으므로, 두 분포 사이에 경계값(threshold)을 설정하면 효과적인 이진 분류가 가능하다. PI는 분리 거리와 분산을 **동시에 비율로 최적화**하기 때문에 효과적이다.

## 문제 2. 생성적 접근 방식으로 $p(C_1|\mathbf{x})$를 시그모이드 형태로 유도 (5점)


### ① 핵심 조건

- 두 클래스가 각각 다변량 정규분포를 따르고, 공분산 행렬 $\Sigma$는 동일하다.
- 이 조건에서 로그 우도비는 $\mathbf{x}$의 이차항이 서로 취소되어 선형식이 된다.


### ② 베이즈 정리와 시그모이드 연결

$p(C_1|\mathbf{x}) = \frac{p(\mathbf{x}|C_1)p(C_1)}{p(\mathbf{x}|C_1)p(C_1)+p(\mathbf{x}|C_0)p(C_0)} = \frac{1}{1+e^{-a}}$

여기서 $a = \ln \frac{p(\mathbf{x}|C_1)p(C_1)}{p(\mathbf{x}|C_0)p(C_0)}$.
$a$는 스칼라여야 시그모이드 함수가 된다.


### ③ 동일 공분산에서 선형식 등장

동일한 공분산 $\Sigma$를 쓸 때, 두 클래스의 로그 확률에서:
- 공통 상수항은 사라지고
- $\mathbf{x}$의 이차항 $\mathbf{x}^T\Sigma^{-1}\mathbf{x}$도 서로 취소된다.
이로써 $a$는 선형식이 된다.


### ④ 최종 결론

$p(C_1|\mathbf{x}) = \sigma(\mathbf{w}^T\mathbf{x} + w_0) = \frac{1}{1+e^{-\mathbf{w}^T\mathbf{x}-w_0}}$

여기서:
- $\mathbf{w} = \Sigma^{-1}(\boldsymbol{\mu}_1 - \boldsymbol{\mu}_0)$
- $w_0 = -\frac{1}{2}\boldsymbol{\mu}_1^T\Sigma^{-1}\boldsymbol{\mu}_1 + \frac{1}{2}\boldsymbol{\mu}_0^T\Sigma^{-1}\boldsymbol{\mu}_0 + \ln\frac{p(C_1)}{p(C_0)}$


### ⑤ 요약

- 시그모이드 형태가 되려면 $a$가 스칼라여야 한다.
- 동일 공분산 조건으로 $a$가 $\mathbf{w}^T\mathbf{x}+w_0$로 나와 로지스틱 모델과 일치한다.
